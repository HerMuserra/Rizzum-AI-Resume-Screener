```python

# Example resume.txt
resume_text = """Experienced Python developer skilled in machine learning, NLP, and data analysis. Worked with TensorFlow, scikit-learn, and Pandas on multiple projects."""
with open("resume.txt", "w") as f:
    f.write(resume_text)

# Example job_description.txt
job_text = """We are looking for a Python developer with experience in machine learning, data analysis, and NLP. Must be familiar with TensorFlow and scikit-learn."""
with open("job_description.txt", "w") as f:
    f.write(job_text)




import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Download stopwords if not already
nltk.download('stopwords')

def load_text(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read()

def preprocess(text):
    stop_words = set(stopwords.words('english'))
    words = text.lower().split()
    words = [w for w in words if w.isalpha() and w not in stop_words]
    return ' '.join(words)

# Load files
resume = preprocess(load_text('resume.txt'))
job_desc = preprocess(load_text('job_description.txt'))

# Vectorize and compute similarity
vectorizer = CountVectorizer()
vectors = vectorizer.fit_transform([resume, job_desc])
similarity = cosine_similarity(vectors)[0][1]

print(f"Rizzumé Match Score: {similarity*100:.2f}%")
```
